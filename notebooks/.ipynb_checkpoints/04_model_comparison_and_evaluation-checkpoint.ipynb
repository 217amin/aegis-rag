# 04 — Model Comparison and Evaluation

Thin demo layer over `scripts/eval.py`.

Compares baseline vs fine-tuned model using the same hybrid RAG pipeline and gold datasets.

> **What changed from the original:**  
> - All constants are imported from `scripts/config` (no stale local copies).  
> - A single `Retriever` instance is shared across all eval calls.  
> - `evaluate_generation_for_model` now takes a `retriever=` kwarg instead of creating resources internally.  
> - RAGAS is skipped by default for the Ollama judge (pass `skip_ragas=False` to enable).

```python
from pathlib import Path
import json, sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
# Config is the single source of truth — no more copy-pasted constants.
from scripts.config import (
    EVAL_DIR,
    LLM_MODEL, FINETUNED_MODEL,
    RESULTS_RETRIEVAL_FILE, RESULTS_GENERATION_FILE, RESULTS_SUMMARY_FILE,
)
import scripts.eval as eval_mod

{
    "EVAL_DIR"                : str(EVAL_DIR),
    "BASE_MODEL"              : LLM_MODEL,
    "FINETUNED_MODEL"         : FINETUNED_MODEL,
    "RESULTS_RETRIEVAL_FILE"  : str(RESULTS_RETRIEVAL_FILE),
    "RESULTS_GENERATION_FILE" : str(RESULTS_GENERATION_FILE),
    "RESULTS_SUMMARY_FILE"    : str(RESULTS_SUMMARY_FILE),
}
```

## Check evaluation inputs

```python
retrieval_gold = eval_mod.load_json(EVAL_DIR / "retrieval_gold.json")
qa_gold        = eval_mod.load_json(EVAL_DIR / "qa_gold.json")

print("retrieval_gold:", len(retrieval_gold), "  qa_gold:", len(qa_gold))
```

```python
retrieval_gold[:2], qa_gold[:2]
```

## Build the shared retriever

Built once here and passed into every eval call — avoids reloading the embedding model and BM25 index per question.

```python
from scripts.retriever import Retriever

retriever = Retriever()
print("Retriever ready.")
```

## Evaluate retrieval

```python
print("Evaluating retrieval...")
retrieval_results = eval_mod.evaluate_retrieval(retrieval_gold, retriever)
print(json.dumps(retrieval_results["aggregate"], indent=2))
```

## Evaluate generation — base model

```python
from scripts.config import LLM_MODEL, JUDGE_MODEL

# We removed judge_provider entirely. 
# We now rely purely on JUDGE_MODEL ("llama3.1:8b") enforcing strict JSON.
base_results = eval_mod.evaluate_generation_for_model(
    qa_gold=qa_gold,
    model_name=LLM_MODEL,
    judge_model=JUDGE_MODEL, # <--- Points to the 8B model we set in config.py
    retriever=retriever,
    request_timeout=300,     # Increased to 300s to give the 8B model time to think
    generation_retries=2,
    ragas_retries=2,
    skip_ragas=False,        # <--- Set to False so we actually run the evaluation!
)

print(json.dumps(base_results["aggregate"], indent=2))
```

## Evaluate generation — fine-tuned model

```python
from scripts.config import LLM_MODEL, JUDGE_MODEL

# We removed judge_provider entirely. 
# We now rely purely on JUDGE_MODEL ("llama3.1:8b") enforcing strict JSON.
base_results = eval_mod.evaluate_generation_for_model(
    qa_gold=qa_gold,
    model_name=FINETUNED_MODEL,
    judge_model=JUDGE_MODEL, # <--- Points to the 8B model we set in config.py
    retriever=retriever,
    request_timeout=300,     # Increased to 300s to give the 8B model time to think
    generation_retries=2,
    ragas_retries=2,
    skip_ragas=False,        # <--- Set to False so we actually run the evaluation!
)

print(json.dumps(base_results["aggregate"], indent=2))
```
