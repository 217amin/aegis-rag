# 02 — Retrieval A/B Testing, Abstention Eval, and Automatic Plots

This notebook runs retrieval-only benchmarking across all configured profiles, evaluates abstention precision on out-of-scope queries, and generates comparison plots automatically.

Profiles come from `scripts.config.RETRIEVAL_PROFILES`:
- `bm25_only`
- `dense_only`
- `hybrid`
- `hybrid_rerank`

Three eval tracks:
1. **Retrieval** — recall@k, nDCG@5, MRR on in-scope queries (`retrieval_gold.json`)
2. **Abstention** — score-gate precision on out-of-scope queries (`abstention_gold.json`)
3. **Qualitative** — eyeball check on a representative compliance query

Recommended order:
1. run `01_data_ingestion_and_chunking.ipynb`
2. run this notebook
3. continue to `03_model_comparison_and_evaluation.ipynb`


```python

from pathlib import Path
import sys
import pandas as pd
from IPython.display import Image, display

PROJECT_ROOT = Path.cwd().resolve().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
from scripts.config import (
    EVAL_DIR,
    PLOTS_DIR,
    RESULTS_ABLATION_FILE,
    RESULTS_ABSTENTION_FILE,
    RETRIEVAL_PROFILES,
)
import scripts.eval as eval_mod

retrieval_gold = eval_mod.load_json(EVAL_DIR / "retrieval_gold.json")
qa_gold = eval_mod.load_json(EVAL_DIR / "qa_gold.json")

abstention_path = EVAL_DIR / "abstention_gold.json"
abstention_gold = eval_mod.load_json(abstention_path) if abstention_path.exists() else []

print("retrieval_gold examples :", len(retrieval_gold))
print("qa_gold examples        :", len(qa_gold))
print("abstention_gold examples:", len(abstention_gold))
print("profiles                :", list(RETRIEVAL_PROFILES.keys()))
```

## Run retrieval A/B testing

```python
from scripts.retriever import load_retrieval_assets
import pandas as pd

print("Loading functional retrieval assets into memory...")
assets = load_retrieval_assets()

print("Running retrieval ablation study...")
ablation_results = [
    eval_mod.evaluate_retrieval_profile(retrieval_gold, profile_name, assets)
    for profile_name in RETRIEVAL_PROFILES.keys()
]

eval_mod.save_json(ablation_results, RESULTS_ABLATION_FILE)
eval_mod.plot_retrieval_profiles(ablation_results, PLOTS_DIR)

ablation_df = pd.DataFrame([
    {"profile_name": item["profile_name"], **item["aggregate"]}
    for item in ablation_results
]).sort_values(["ndcg@5", "recall@5"], ascending=False)

print("Saved ablation results →", RESULTS_ABLATION_FILE)
ablation_df
```

## Generated plots

```python
display(Image(filename=str(PLOTS_DIR / "retrieval_profile_ndcg.png")))
display(Image(filename=str(PLOTS_DIR / "retrieval_profile_recall.png")))
display(Image(filename=str(PLOTS_DIR / "latency_quality_tradeoff.png")))
```

## Inspect one profile in detail

```python
best_profile = "hybrid_rerank"
profile_detail = next(x for x in ablation_results if x["profile_name"] == best_profile)

print("Forced Best Profile (MRM Approved):", best_profile)

# Display the per-query breakdown for the chosen profile
df_detail = pd.DataFrame(profile_detail["per_query"])[
    ["query", "recall@5", "ndcg@5", "mrr", "retrieved_ids", "relevant_ids"]
]

df_detail.head(10)
```

## Run abstention A/B testing

For each out-of-scope query, run retrieval and check whether the score-gate (`top_rerank_score < NO_ANSWER_MIN_RERANK_SCORE` for rerank profiles, `top_weighted_rrf_score < NO_ANSWER_MIN_FUSED_SCORE` for fused-only) would fire.

A correct abstention = the gate fires. Precision = correct / applicable.


```python
import pandas as pd

if abstention_gold:
    print("Running abstention evaluation...")
    abstention_results = [
        # PASS ASSETS HERE:
        eval_mod.evaluate_abstention(abstention_gold, profile_name, assets)
        for profile_name in RETRIEVAL_PROFILES.keys()
    ]
    eval_mod.save_json(abstention_results, RESULTS_ABSTENTION_FILE)

    abstention_df = pd.DataFrame([
        {"profile_name": item["profile_name"], **item["aggregate"]}
        for item in abstention_results
    ])

    print("Saved abstention results ->", RESULTS_ABSTENTION_FILE)
    display(abstention_df)

    print("\nPer-query detail (hybrid_rerank):")
    rerank_detail = next(
        (r for r in abstention_results if r["profile_name"] == "hybrid_rerank"),
        None,
    )
    if rerank_detail:
        display(pd.DataFrame(rerank_detail["per_query"])[[
            "question", "category", "gate_score", "gate_threshold", "abstained_correctly"
        ]])
else:
    print("abstention_gold.json not found in EVAL_DIR -- skipping abstention eval.")
    abstention_results = []
```

## Quick qualitative retrieval check

```python
from scripts.retriever import load_retrieval_assets, retrieve_documents
from scripts.config import RETRIEVAL_PROFILES

print("Loading functional retrieval assets...")
assets = load_retrieval_assets()

# In-scope query — should retrieve relevant FINTRAC chunks with high rerank score
test_query = "What is the 24-hour rule for cash transactions?"
print(f"IN-SCOPE QUERY: {test_query!r}\n")

for profile_name in RETRIEVAL_PROFILES.keys():
    docs, diagnostics = retrieve_documents(test_query, assets, profile_name)
    print(f"PROFILE: {profile_name}")
    print("  rewritten_query :", diagnostics.get("rewritten_query"))
    print("  top_rerank_score:", diagnostics.get("top_rerank_score"))
    print("  top_fused_score :", diagnostics.get("top_weighted_rrf_score"))
    # Note: 'docs' are now dicts, not LangChain objects
    print("  chunk_ids       :", [d.get("chunk_id") for d in docs[:3]])
    print()

# Out-of-scope query — score-gate should fire on hybrid_rerank
test_query_oos = "What time is standard hotel check-in?"
print(f"\nOUT-OF-SCOPE QUERY: {test_query_oos!r}\n")

for profile_name in ["hybrid", "hybrid_rerank"]:
    docs, diagnostics = retrieve_documents(test_query_oos, assets, profile_name)
    print(f"PROFILE: {profile_name}")
    print("  top_rerank_score:", diagnostics.get("top_rerank_score"))
    print("  top_fused_score :", diagnostics.get("top_weighted_rrf_score"))
    print("  chunk_ids       :", [d.get("chunk_id") for d in docs[:3]])
    print()
```
